```python
import numpy as np
import pandas as pd
```

```python
#create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
#Inspection and Properties
print("1D Array Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
#Array Operations
fahrenheit_w1 = (temperatures_w1 * 9/5) + 32
print("\nTemperatures in Fahrenheit:", fahrenheit_w1)
print("Max Temp:", np.max(temperatures_w1))
print("Min Temp:", np.min(temperatures_w1))
print("Mean Temp:", np.mean(temperatures_w1))
```

```python
#Arrary Slicing and Indexing
print("\nFirst 3 days:", temperatures_w1[:3])
print("Weekend (Last 2 days):", temperatures_w1[-2])
print("Middle 3 days:", temperatures_w1[2:5])
```

```python
#Create a 2D Array
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9], [19.2, 22.5,
21.3, 24.0, 23.5, 22.8, 20.1]])
```

```python
#Inspect and Slice the 2D Array
print("\n2D Array shape:",
temperatures.shape)
print("2D Data Type:",
temperatures.dtype)
print("2D Total Elements:", temperatures.size)
```

```python
print("Week 1 Temperatures:", temperatures[0])
print("Week 2 Temperatures:", temperatures[1])
print("Weekend (Last 2 days for both weeks):\n", temperatures[:, -2:])
```

```python
import pandas as pd
#Creating pandas series
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3',
'Rank4', 'Rank5' ])
print("Original Series:\n", marks)
```

```python
#Indexing and Slicing
print("\n1st rank student mark (position):", marks.iloc[0])
print("Top 3 ranks:\n", marks.loc['Rank1':'Rank3'])
print("3rd-ranked student mark:", marks.iloc[2])
print("Rank with marks >90:\n", marks[marks>90])
```

```python
#Manipulating Series
marks['Rank1'] = 100
marks = marks.drop('Rank5')
cgpa =marks / 10
print("\nModified Series:\n", marks)
print("cgpa:\n", cgpa)
```

```python
import pandas as pd
#Creating pandas DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
    'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]

  }
transactions = pd.DataFrame(data)
```

```python
#Data Exloration
print("---DataFrame Output---")
print(transactions)
print("\n---Head---")
print(transactions.head())
print("\n---Tail---")
print(transactions.tail())
print("\n---Shape---")
print(transactions.shape)
print("\n---Info---")
transactions.info()
```

```python
#ProductCategory & Amount columns
print("\n, transactions[['ProductCategory','Amount']]")
```

```python
#Last 3 columns
print("\nLast 3 columns:\n", transactions.iloc[:,-3:])
```

```python
#Filter: Region == 'North' and Amount >200
filtered_df =transactions[(transactions['Region'] == 'North') & (transactions['Amount']>200)]
print("\nFiltered_df")
```

```python
#Value counts & Unique values
print("\nProductCategory Count:\n" , transactions['ProductCategory'].value_counts())
print("\nUnique Regions:", transactions['Region'].unique())
```

```python
#Groupby Region Mean Amount
print("\nMean Amount by Region:\n", transactions.groupby('Region')['Amount'].mean())
```

```python
#Manipulation DataFrame
#Modify Amount for TransactionID 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] =165
```

```python
#Remove row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]
```

```python
#Create Discount column first
transactions['Discount'] = transactions['Amount'] * 0.10
#Delete 'Discount' column
transactions =transactions.drop(columns=['Discount'])
print("\n--Final DataFrame---")
print(transactions)
```
